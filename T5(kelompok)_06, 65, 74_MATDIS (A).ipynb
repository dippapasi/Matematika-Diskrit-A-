```python
import networkx as nx
import matplotlib.pyplot as plt

# Inisialisasi Graph
G = nx.Graph()

# Fungsi Menggambar Graph
def draw_graph(graph, title="Graf Pertemanan Mahasiswa"):
    plt.figure(figsize=(14, 10))
    pos = nx.kamada_kawai_layout(graph)

    node_colors = [data["color"] for _, data in graph.nodes(data=True)]
    edge_colors = [data["color"] for _, _, data in graph.edges(data=True)]

    nx.draw(
        graph,
        pos,
        with_labels=True,
        node_size=800,
        font_size=8,
        font_color="white",
        node_color=node_colors,
        edge_color=edge_colors,
        width=2,
        alpha=0.9,
    )

    plt.title(title, fontsize=15, fontweight="bold")

    # Legend Warna Klaster
    for label, color in legend_colors.items():
        plt.scatter([], [], color=color, label=label, s=150, edgecolors="black")

    legend = plt.legend(
        title="Keterangan Klaster",
        loc="upper right",
        fontsize=9,
        title_fontsize=10
    )
    legend.get_frame().set_alpha(0.9)

    plt.axis("off")
    plt.subplots_adjust(right=0.8)
    plt.show()

# Data Mahasiswa
mahasiswa = [
    "Madina", "Maria", "Adinda", "Aquina", "Alimun", "Hizkia", "Raveena",
    "Ophyng", "Aisyah", "Febriani", "Amelia", "Via", "Fanny", "Naia",
    "Alfani", "Diah", "Kenzy", "Angel", "Maulida", "Izzati", "Auliya",
    "Carissa", "Diva", "Erik", "Maya", "Dany", "Gendis", "Wildan", "Esthi",
    "Naufal", "Laudya", "Adrian", "Selvy", "Hafiyyan", "Indra", "Yuniar",
    "Kiky", "Sofia", "Varid", "Gaitsa"
]
G.add_nodes_from(mahasiswa)

# Klaster Awal
sekelompok = [
    ("Raveena", "Amelia"), ("Raveena", "Yuniar"), ("Yuniar", "Amelia"),
    ("Hizkia", "Maria"), ("Hizkia", "Sofia"), ("Sofia", "Maria"),
    ("Naufal", "Laudya"), ("Naufal", "Febriani"), ("Febriani", "Laudya"),
    ("Kenzy", "Wildan"), ("Wildan", "Hafiyyan"), ("Kenzy", "Hafiyyan"),
    ("Alimun", "Dany"), ("Dany", "Indra"), ("Indra", "Alimun"),
    ("Esthi", "Kiky"), ("Kiky", "Selvy"), ("Selvy", "Esthi"),
    ("Adrian", "Izzati"), ("Izzati", "Fanny"), ("Fanny", "Adrian"),
    ("Aquina", "Diva"), ("Diva", "Maya"), ("Maya", "Aquina"),
    ("Diah", "Gendis"), ("Gendis", "Angel"), ("Angel", "Diah"),
    ("Ophyng", "Aisyah"), ("Aisyah", "Naia"), ("Naia", "Ophyng"),
    ("Adinda", "Varid"), ("Varid", "Gaitsa"), ("Gaitsa", "Adinda"),
    ("Auliya", "Carissa"), ("Carissa", "Erik"), ("Erik", "Auliya"),
    ("Madina", "Via"), ("Maulida", "Madina"), ("Via", "Alfani"),
    ("Madina", "Alfani"), ("Maulida", "Via"), ("Maulida", "Alfani")
]
G.add_edges_from(sekelompok, color="red")

# Fungsi Tambah Klaster
def tambah_klaster(anggota, warna):
    for i in range(len(anggota)):
        for j in range(i + 1, len(anggota)):
            G.add_edge(anggota[i], anggota[j], color=warna)

# Klaster Tambahan
akamsi = ["Yuniar", "Via", "Raveena", "Kenzy", "Esthi", "Alimun", "Amelia",
          "Indra", "Madina", "Maulida", "Kiky", "Wildan"]
jabodetabek = ["Sofia", "Maria", "Fanny", "Gaitsa", "Aisyah"]
platP = ["Angel", "Yuniar"]
platS = ["Naia", "Izzati"]
platM = ["Dany", "Alimun"]
platN = ["Amelia", "Alfani"]
platAx = ["Adrian", "Raveena", "Selvy", "Fanny"]
sekos = ["Aisyah", "Naia", "Diah", "Fanny"]

himasada = ["Dany", "Yuniar", "Kiky", "Esthi", "Raveena",
            "Wildan", "Selvy", "Via", "Madina", "Alfani", "Maulida"]
BEMfakultas = ["Indra", "Hafiyyan", "Varid", "Carissa", "Raveena"]
warkopbening = ["Raveena", "Yuniar", "Wildan", "Dany", "Maulida", "Izzati"]
skincaredanmakeup = ["Hizkia", "Kiky", "Yuniar", "Raveena", "Sofia", "Amelia"]

drakor_lovers = ["Naia", "Aisyah", "Sofia", "Maria", "Gendis", "Aquina", "Maya", "Yuniar"]
gaming_team = ["Wildan", "Ophyng", "Indra", "Erik", "Kenzy", "Diah", "Aquina"]
pecinta_kopi = ["Raveena", "Yuniar", "Madina", "Maulida", "Dany", "Maya", "Gendis"]
tugas_bareng = ["Adinda", "Varid", "Gaitsa", "Carissa", "Auliya", "Raveena", "Wildan", "Angel", "Diva"]

# Warna Klaster
klaster_dan_warna = [
    (akamsi, "orange"), (jabodetabek, "lightblue"), (platP, "violet"),
    (platS, "gray"), (platM, "green"), (platN, "yellow"),
    (platAx, "red"), (sekos, "lightgreen"),
    (himasada, "gold"), (BEMfakultas, "purple"),
    (warkopbening, "lightcoral"), (skincaredanmakeup, "pink"),
    (drakor_lovers, "cyan"), (gaming_team, "blue"),
    (pecinta_kopi, "brown"), (tugas_bareng, "black")
]

for klaster, warna in klaster_dan_warna:
    tambah_klaster(klaster, warna)

# Legend Warna Klaster
legend_colors = {
    "Akamsi": "orange",
    "Jabodetabek": "lightblue",
    "Plat P": "violet",
    "Plat S": "gray",
    "Plat M": "green",
    "Plat N": "yellow",
    "Plat Ax": "red",
    "Sekos": "lightgreen",
    "Himasada": "gold",
    "BEM Fakultas": "purple",
    "Warkop Bening": "lightcoral",
    "Skincare & Makeup": "pink",
    "Drakor Lovers": "cyan",
    "Gaming Team": "blue",
    "Pecinta Kopi": "brown",
    "Tugas Bareng": "black"
}

# Warna Node
degrees = dict(G.degree())
max_deg = max(degrees.values())

for node, deg in degrees.items():
    if deg >= 0.75 * max_deg:
        color = "lightgreen"      # sangat populer
    elif deg >= 0.5 * max_deg:
        color = "gold"            # sedang
    else:
        color = "lightcoral"      # sedikit koneksi
    G.nodes[node]["color"] = color

# Visualisasi Graph
draw_graph(G, "Jaringan Pertemanan Mahasiswa")
```

#**Analisis 2 : mencari pasangan - pasangan yang memilkiki koneksi terjauh**

```python
import networkx as nx
import matplotlib.pyplot as plt

# Analisis No. 2: Pasangan dengan koneksi paling jauh + Visualisasi

# Cek konektivitas graph
if not nx.is_connected(G):
    largest_cc = max(nx.connected_components(G), key=len)
    G_sub = G.subgraph(largest_cc).copy()
    print("Graf tidak sepenuhnya terhubung, menggunakan komponen terbesar.")
else:
    G_sub = G

# Hitung diameter graph (jarak koneksi paling jauh)
diameter = nx.diameter(G_sub)
print(f"Jarak koneksi terpanjang (diameter): {diameter}")

# Cari semua pasangan yang jaraknya sama dengan diameter
farthest_pairs = []
for u in G_sub.nodes():
    for v in G_sub.nodes():
        if u != v:
            try:
                dist = nx.shortest_path_length(G_sub, u, v)
                if dist == diameter:
                    farthest_pairs.append((u, v))
            except nx.NetworkXNoPath:
                pass

# Hapus duplikat pasangan
unique_pairs = set(tuple(sorted(p)) for p in farthest_pairs)
print("\nPasangan dengan koneksi paling jauh:")
for p in unique_pairs:
    print(p)

# Pilih satu pasangan untuk divisualisasikan
pair = list(unique_pairs)[0]
path = nx.shortest_path(G_sub, source=pair[0], target=pair[1])
print(f"\nRute terpendek antara {pair[0]} dan {pair[1]}: {path}")

# Visualisasi path terpanjang
pos = nx.kamada_kawai_layout(G_sub)

# Warna default edge
edge_colors = []
for u, v in G_sub.edges():
    if any((u == path[i] and v == path[i+1]) or (v == path[i] and u == path[i+1]) for i in range(len(path)-1)):
        edge_colors.append("red")  # jalur terpanjang ditandai merah
    else:
        edge_colors.append("lightgray")

plt.figure(figsize=(14, 10))
nx.draw_networkx_nodes(G_sub, pos, node_color="lightblue", node_size=700, alpha=0.9)
nx.draw_networkx_labels(G_sub, pos, font_size=8, font_color="black")
nx.draw_networkx_edges(G_sub, pos, edge_color=edge_colors, width=2)

# Sorot node awal dan akhir
nx.draw_networkx_nodes(G_sub, pos, nodelist=[path[0]], node_color="green", node_size=1000, label=f"Awal: {path[0]}")
nx.draw_networkx_nodes(G_sub, pos, nodelist=[path[-1]], node_color="orange", node_size=1000, label=f"Akhir: {path[-1]}")

plt.title(f"Visualisasi Pasangan Terjauh ({pair[0]} - {pair[1]})\nJarak = {diameter}", fontsize=14, fontweight="bold")
plt.legend(loc="upper right")
plt.axis("off")
plt.show()
```

**Interpretasi Hasil Output Analisis 2**

Visualisasi jaringan di atas menunjukkan seluruh pasangan dengan jarak koneksi terjauh dalam graf. Terlihat bahwa setiap pasangan seperti *Erik–Hizkia*, *Adrian–Ophyng*, *Hafiyyan–Naia*, dan lainnya memiliki lintasan unik yang dihubungkan oleh simpul-simpul perantara sesuai hasil perhitungan *shortest path*. Warna yang berbeda pada tiap jalur menandakan rute antar pasangan dengan jarak maksimal (diameter graf), sehingga memudahkan untuk mengamati distribusi keterhubungan antar individu. Node berwarna lebih mencolok menandai titik awal dan akhir dari tiap pasangan, menunjukkan simpul-simpul yang berperan penting dalam jangkauan komunikasi terjauh jaringan. Semakin panjang lintasan dan semakin beragam simpul perantara yang dilewati, semakin kompleks hubungan dalam graf tersebut. Secara keseluruhan, visualisasi ini memperlihatkan bahwa meskipun jaringan terhubung, beberapa pasangan membutuhkan banyak langkah untuk saling mencapai, menandakan adanya kelompok atau cluster yang relatif terpisah dalam struktur hubungan.


#**Analisis 3 : bridge person**
a) Bridge person. nx.betweenness_centrality(G)

b) Cari bridge person dengan koneksi paling sedikit

```python
# Analisis 3a: bridge person
betweenness = nx.betweenness_centrality(G)
bridge = sorted(betweenness.items(), key=lambda x: x[1], reverse=True)[:3]

print("Top 3 Bridge Person di Jaringan Pertemanan")
for i, (n, v) in enumerate(bridge, 1):
    print(f"{i}. {n} - {v:.4f}")

top_nodes = [n for n, _ in bridge]
pos = nx.kamada_kawai_layout(G)

plt.figure(figsize=(12,8))
nx.draw(G, pos, node_color="lightgray", edge_color="lightgray", node_size=400, alpha=0.6)
nx.draw_networkx_nodes(G, pos, nodelist=top_nodes, node_color="orange", node_size=1000, label="Bridge Person", edgecolors="black", linewidths=2)
nx.draw_networkx_labels(G, pos, font_size=8)
plt.title("Bridge Person di Jaringan Pertemanan", fontsize=13, fontweight="bold")
plt.legend(); plt.axis("off"); plt.show()
```

**Interpretasi Hasil Output Analisis 3a:**

Berdasarkan hasil analisis *betweenness centrality*, tiga mahasiswa dengan nilai tertinggi yaitu Raveena (0.14), Wildan (0.11), dan Yuniar (0.07) berperan sebagai *bridge person* atau penghubung utama dalam jaringan pertemanan. Nilai tersebut sejalan dengan visualisasi, di mana ketiganya nampak memiliki banyak garis koneksi yang menghubungkkan berbagai kelompok, menandakan posisi strategis mereka dalam menjembatani interaksi antar teman.

```python
# Analisis 3b: bridge person dengan koneksi sedikit
import pandas as pd

koneksi = dict(G.degree())
betweenness = nx.betweenness_centrality(G)

df_bridge = pd.DataFrame({
    "Nama": list(G.nodes()),
    "Koneksi": [koneksi[n] for n in G.nodes()],
    "Betweenness": [betweenness[n] for n in G.nodes()]
})

# Diurutkan dari koneksi paling sedikit
df_bridge = df_bridge.sort_values(by=["Koneksi", "Betweenness"], ascending=[True, False]).reset_index(drop=True)

print("Orang dengan koneksi sedikit tapi berperan sebagai penghubung penting:\n")
df_bridge.head()
```

**Interpretasi Hasil Ouput Analisis 3b:**

Berdasarkan hasil analisis dengan menampilkan lima mahasiswa teratas dengan koneksi paling sedikit, diketahui bahwa Febriani, Naufal, dan Laudya memiliki jumlah koneksi sangat sedikit (masing-masing 2 koneksi) serta nilai betweenness sebesar 0, yang menunjukkan bahwa mereka belum berperan sebagai penghubung antar kelompok dan hanya berinteraksi dalam lingkaran sosial terbatas. Sementara itu, Adrian dan Hizkia memiliki koneksi yang relatif sedikit (4 dan 6 koneksi) namun nilai betweenness lebih tinggi dibandingkan lainnya, menandakan bahwa keduanya mulai berperan sebagai *bridge person*, yaitu penghubug yang menjaga keterhubungan antara beberapa kelompok dalam jaringan pertemanan mahasiswa. Hal ini menunjukkan bahwa jumlah koneksi yang sedikit tidak selalu berarti tidak berpengaruh, karena posisi seseorang dalam jaringan dapat tetap strategis meskipun dengan relasi yang terbatas.
